```python
import itertools

# 1. 定义 D4 群的 8 种对称变换
def get_symmetries(board):
    """返回棋盘 board 的所有 8 个对称变换局面"""
    # 将一维转为二维方便操作
    b = [board[i:i+3] for i in range(0, 9, 3)]
    
    def rotate(m): return [list(r) for r in zip(*m[::-1])]
    def flip(m): return [r[::-1] for r in m]
    
    syms = []
    curr = b
    for _ in range(4): # 4个旋转
        curr = rotate(curr)
        syms.append(tuple(itertools.chain(*curr)))
        syms.append(tuple(itertools.chain(*flip(curr)))) # 每个旋转后的翻转
    return set(syms)

def get_canonical(board):
    """返回该局面所在轨道的代表元（字典序最小的状态）"""
    return min(get_symmetries(board))

def solve():
    # 初始状态：全 0 棋盘 
    initial_board = (0,) * 9
    
    # 存储每一层的本质不同状态
    unique_states_per_turn = {i: set() for i in range(10)}
    unique_states_per_turn[0].add(initial_board)
    
    # 记录原始状态数用于对比
    raw_states_per_turn = {i: set() for i in range(10)}
    raw_states_per_turn[0].add(initial_board)

    print(f"{'步数(t)':<8} | {'原始状态数':<12} | {'约简后状态数(轨道数)':<15} | {'约简率':<10}")
    print("-" * 60)

    for t in range(9):
        # 确定当前轮到谁下
        current_player = 1 if t % 2 == 0 else -1
        
        for board in raw_states_per_turn[t]:
            # 在每一个空位尝试下棋
            for i in range(9):
                if board[i] == 0:
                    new_board = list(board)
                    new_board[i] = current_player
                    new_board = tuple(new_board)
                    
                    # 记录原始状态
                    raw_states_per_turn[t+1].add(new_board)
                    
                    # 记录约简后的代表元
                    unique_states_per_turn[t+1].add(get_canonical(new_board))
        
        # 计算并输出当前层的统计数据
        raw_count = len(raw_states_per_turn[t+1])
        reduced_count = len(unique_states_per_turn[t+1])
        reduction_rate = (1 - reduced_count / raw_count) * 100
        
        print(f"t={t+1:<6} | {raw_count:<15} | {reduced_count:<22} | {reduction_rate:.1f}%")

if __name__ == "__main__":
    solve()
```
